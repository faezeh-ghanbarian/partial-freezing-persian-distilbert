```python
import conllu

with open("fa_perdt-ud-train.conllu", "r", encoding="utf-8") as f:
    train_data = list(conllu.parse(f.read()))

print(len(train_data))
print(train_data[0][:5])
```

```python
id2="extract_pos"
def extract_tokens_labels(data):
    sentences = []
    labels = []

    for sentence in data:
        tokens = []
        pos_tags = []

        for token in sentence:
            tokens.append(token["form"])
            pos_tags.append(token["upos"])

        sentences.append(tokens)
        labels.append(pos_tags)

    return sentences, labels


train_tokens, train_labels = extract_tokens_labels(train_data)

print(train_tokens[0])
print(train_labels[0])
```

```python
from collections import Counter

all_labels = [label for sent in train_labels for label in sent]

label_list = sorted(list(set(all_labels)))

label2id = {label: i for i, label in enumerate(label_list)}
id2label = {i: label for label, i in label2id.items()}

print(label_list)
print(label2id)
```

```python
with open("fa_perdt-ud-dev.conllu", "r", encoding="utf-8") as f:
    dev_data = list(conllu.parse(f.read()))

with open("fa_perdt-ud-test.conllu", "r", encoding="utf-8") as f:
    test_data = list(conllu.parse(f.read()))

dev_tokens, dev_labels = extract_tokens_labels(dev_data)
test_tokens, test_labels = extract_tokens_labels(test_data)

print(len(dev_tokens))
print(len(test_tokens))
```

```python
from transformers import AutoTokenizer

model_checkpoint = "distilbert-base-multilingual-cased"

tokenizer = AutoTokenizer.from_pretrained(model_checkpoint)

example = tokenizer(
    train_tokens[0],
    is_split_into_words=True
)

print(example.tokens()[:20])
print(example.word_ids()[:20])
```

```python
def tokenize_and_align_labels(tokens, labels):
    tokenized_inputs = tokenizer(
        tokens,
        truncation=True,
        is_split_into_words=True,
        max_length=128
    )

    word_ids = tokenized_inputs.word_ids()

    aligned_labels = []

    previous_word_id = None

    for word_id in word_ids:
        if word_id is None:
            aligned_labels.append(-100)
        elif word_id != previous_word_id:
            aligned_labels.append(label2id[labels[word_id]])
        else:
            aligned_labels.append(-100)

        previous_word_id = word_id

    tokenized_inputs["labels"] = aligned_labels

    return tokenized_inputs
```

```python
sample = tokenize_and_align_labels(
    train_tokens[0],
    train_labels[0]
)

print(sample["input_ids"][:20])
print(sample["labels"][:20])
```

```python
from datasets import Dataset

train_encodings = [
    tokenize_and_align_labels(tokens, labels)
    for tokens, labels in zip(train_tokens, train_labels)
]

dev_encodings = [
    tokenize_and_align_labels(tokens, labels)
    for tokens, labels in zip(dev_tokens, dev_labels)
]

test_encodings = [
    tokenize_and_align_labels(tokens, labels)
    for tokens, labels in zip(test_tokens, test_labels)
]

train_dataset = Dataset.from_list(train_encodings)
dev_dataset = Dataset.from_list(dev_encodings)
test_dataset = Dataset.from_list(test_encodings)

print(train_dataset)
```

```python
from transformers import (
    AutoModelForTokenClassification,
    Trainer,
    TrainingArguments,
)

model_checkpoint = "distilbert-base-multilingual-cased"

model = AutoModelForTokenClassification.from_pretrained(
    model_checkpoint,
    num_labels=len(label_list),
    id2label=id2label,
    label2id=label2id
)

print(model.num_labels)
```

```python
from transformers import DataCollatorForTokenClassification

data_collator = DataCollatorForTokenClassification(
    tokenizer=tokenizer
)
```

```python
import numpy as np
from sklearn.metrics import accuracy_score

def compute_metrics(eval_pred):
    predictions, labels = eval_pred

    predictions = np.argmax(predictions, axis=2)

    true_predictions = []
    true_labels = []

    for prediction, label in zip(predictions, labels):
        for p, l in zip(prediction, label):
            if l != -100:
                true_predictions.append(p)
                true_labels.append(l)

    return {
        "accuracy": accuracy_score(
            true_labels,
            true_predictions
        )
    }
```

```python
train_dataset_small = train_dataset.shuffle(seed=42).select(range(8000))
dev_dataset_small = dev_dataset.shuffle(seed=42).select(range(1000))
```

```python
train_dataset_small = train_dataset.shuffle(seed=42).select(range(8000))
dev_dataset_small = dev_dataset.shuffle(seed=42).select(range(1000))

print(train_dataset_small)
print(dev_dataset_small)
```

```python
training_args = TrainingArguments(
    output_dir="./pos_full_finetuning",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    weight_decay=0.01,
    logging_steps=50,
    report_to="none",
    seed=42,
    data_seed=42,
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset_small,
    eval_dataset=dev_dataset_small,
    processing_class=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics
)
```

```python
trainer.train()
```

```python
model_freeze = AutoModelForTokenClassification.from_pretrained(
    model_checkpoint,
    num_labels=len(label_list),
    id2label=id2label,
    label2id=label2id
)
```

```python
for name, param in model_freeze.named_parameters():
    print(name)
    break
```

```python
print(model_freeze.distilbert.transformer.layer)
```

```python
# Freeze embeddings
for param in model_freeze.distilbert.embeddings.parameters():
    param.requires_grad = False

# Freeze first 3 transformer layers
for layer in model_freeze.distilbert.transformer.layer[:3]:
    for param in layer.parameters():
        param.requires_grad = False
```

```python
total_params = sum(p.numel() for p in model_freeze.parameters())

trainable_params = sum(
    p.numel() for p in model_freeze.parameters()
    if p.requires_grad
)

print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
print(f"Trainable ratio: {100*trainable_params/total_params:.2f}%")
```

```python
training_args_freeze = TrainingArguments(
    output_dir="./pos_partial_freezing",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    weight_decay=0.01,
    logging_steps=50,
    report_to="none",
    seed=42,
    data_seed=42,
)

trainer_freeze = Trainer(
    model=model_freeze,
    args=training_args_freeze,
    train_dataset=train_dataset_small,
    eval_dataset=dev_dataset_small,
    processing_class=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics
)
```

```python
trainer_freeze.train()
```

The partial freezing experiment reduced the number of trainable parameters to 15.79% while maintaining a competitive accuracy of 94.13%, compared with 95.73% from full fine-tuning. This suggests that a large portion of the pretrained multilingual representations can be preserved while adapting only selected higher layers for Persian POS tagging.
